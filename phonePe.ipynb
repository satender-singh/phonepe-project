```python
import json
base_path = r"D:\New Volume\DS_GUVI\practise\pulse\data"

transaction_path = os.path.join(
    base_path,
    "aggregated",
    "transaction",
    "country",
    "india",
    "state"
)

total_rows = 0

with engine.begin() as connection:

    for state in os.listdir(transaction_path):

        state_path = os.path.join(transaction_path, state)

        if not os.path.isdir(state_path):
            continue

        for year in os.listdir(state_path):

            year_path = os.path.join(state_path, year)

            if not os.path.isdir(year_path):
                continue

            for quarter_file in os.listdir(year_path):

                if not quarter_file.endswith(".json"):
                    continue

                quarter = int(quarter_file.replace(".json", ""))

                file_path = os.path.join(
                    year_path,
                    quarter_file
                )

                with open(file_path, "r", encoding="utf-8") as f:
                    data = json.load(f)

                transaction_data = data["data"]["transactionData"]

                for item in transaction_data:

                    transaction_type = item["name"]

                    for instrument in item["paymentInstruments"]:

                        transaction_count = instrument["count"]
                        transaction_amount = instrument["amount"]

                        connection.execute(
                            db.text("""
                                INSERT INTO aggregated_transaction
                                (
                                    state,
                                    year,
                                    quarter,
                                    transaction_type,
                                    transaction_count,
                                    transaction_amount
                                )
                                VALUES
                                (
                                    :state,
                                    :year,
                                    :quarter,
                                    :transaction_type,
                                    :transaction_count,
                                    :transaction_amount
                                )
                            """),
                            {
                                "state": state,
                                "year": int(year),
                                "quarter": quarter,
                                "transaction_type": transaction_type,
                                "transaction_count": transaction_count,
                                "transaction_amount": transaction_amount
                            }
                        )

                        total_rows += 1

print("✅ Data inserted successfully!")
print("Total rows inserted:", total_rows)
```

```python
import os
import json
from sqlalchemy import text

base_path = r"C:\Users\ACER\Downloads\data-20260909T111909Z-1-001\data\top\transaction\country\india"

insert_query = text("""
    INSERT INTO top_transaction
    (state, year, quarter, metric_type, transaction_count, transaction_amount)
    VALUES
    (:state, :year, :quarter, :metric_type, :transaction_count, :transaction_amount)
""")

total_rows = 0

with engine.begin() as conn:
    for year in range(2018, 2025):
        year_path = os.path.join(base_path, str(year))

        for filename in sorted(os.listdir(year_path)):
            if not filename.endswith(".json"):
                continue

            quarter = int(filename.replace(".json", ""))
            file_path = os.path.join(year_path, filename)

            with open(file_path, "r", encoding="utf-8") as f:
                json_data = json.load(f)

            states = json_data["data"]["states"]

            for row in states:
                conn.execute(
                    insert_query,
                    {
                        "state": row["entityName"],
                        "year": year,
                        "quarter": quarter,
                        "metric_type": row["metric"]["type"],
                        "transaction_count": row["metric"]["count"],
                        "transaction_amount": row["metric"]["amount"]
                    }
                )

            total_rows += len(states)
            print(f"{year} Q{quarter} → {len(states)} rows")

print(f"\nTotal rows inserted: {total_rows}")
```

```python
import os
import json
from sqlalchemy import text

user_path = r"C:\Users\ACER\Downloads\data-20260909T111909Z-1-001\data\aggregated\user\country\india\state"

insert_query = text("""
    INSERT INTO aggregated_user
    (state, year, quarter, brand, registered_users, app_opens)
    VALUES
    (:state, :year, :quarter, :brand, :registered_users, :app_opens)
""")

total_rows = 0
files_processed = 0
files_without_devices = 0

with engine.begin() as connection:

    for state in os.listdir(user_path):

        state_path = os.path.join(user_path, state)

        if not os.path.isdir(state_path):
            continue

        for year in os.listdir(state_path):

            year_path = os.path.join(state_path, year)

            if not os.path.isdir(year_path):
                continue

            try:
                year_int = int(year)
            except ValueError:
                continue

            for file in os.listdir(year_path):

                if not file.endswith(".json"):
                    continue

                quarter = int(file.replace(".json", ""))
                file_path = os.path.join(year_path, file)

                with open(file_path, "r", encoding="utf-8") as f:
                    data = json.load(f)

                user_data = data.get("data", {})

                aggregated = user_data.get("aggregated") or {}

                app_opens = aggregated.get("appOpens", 0)

                # Handles both a list and None
                devices = user_data.get("usersByDevice") or []

                if not devices:
                    files_without_devices += 1

                for device in devices:

                    brand = device.get("brand")
                    brand_count = device.get("count", 0)

                    connection.execute(
                        insert_query,
                        {
                            "state": state,
                            "year": year_int,
                            "quarter": quarter,
                            "brand": brand,
                            "registered_users": brand_count,
                            "app_opens": app_opens
                        }
                    )

                    total_rows += 1

                files_processed += 1

print("====================================")
print("Data loading completed")
print("Files processed:", files_processed)
print("Files without device data:", files_without_devices)
print("Rows inserted:", total_rows)
print("====================================")
```

```python
#Identify Growth Opportunities
from sqlalchemy import text
import pandas as pd

with engine.connect() as connection:
    df_growth_opportunity = pd.read_sql(text("""
        WITH transactions AS (
            SELECT
                state,
                SUM(transaction_count) AS total_transactions
            FROM map_transaction
            WHERE year = 2025
            GROUP BY state
        ),
        users AS (
            SELECT
                state,
                registered_users
            FROM map_user
            WHERE year = 2025
              AND quarter = 4
        )
        SELECT
            u.state,
            u.registered_users,
            t.total_transactions,

            ROUND(
                1.0 * t.total_transactions /
                NULLIF(u.registered_users, 0),
                2
            ) AS transactions_per_registered_user

        FROM users u
        JOIN transactions t
            ON u.state = t.state

        WHERE u.registered_users >= 20000000

        ORDER BY transactions_per_registered_user ASC;
    """), connection)

df_growth_opportunity
```

```python
#Final Year / Quarter Coverage Validation
from sqlalchemy import text

tables = [
    "aggregated_transaction",
    "aggregated_user",
    "aggregated_insurance",
    "map_transaction",
    "map_user",
    "map_insurance",
    "top_transaction",
    "top_user",
    "top_insurance"
]

with engine.connect() as connection:

    print("=" * 90)
    print("FINAL YEAR / QUARTER COVERAGE VALIDATION")
    print("=" * 90)

    for table in tables:

        query = text(f"""
            SELECT
                MIN(year) AS first_year,
                MAX(year) AS last_year,
                COUNT(DISTINCT year) AS year_count,
                COUNT(DISTINCT quarter) AS quarter_count,
                COUNT(DISTINCT state) AS state_count
            FROM {table}
        """)

        row = connection.execute(query).fetchone()

        print(
            f"{table:25} : "
            f"Years {row.first_year}-{row.last_year} | "
            f"Year count {row.year_count} | "
            f"Quarter count {row.quarter_count} | "
            f"States {row.state_count}"
        )

    print("=" * 90)
```
